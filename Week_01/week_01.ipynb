```python

plt.figure(figsize=(7, 4))
sns.countplot(x='Satisfaction Level', data=df, palette='viridis', order=['Satisfied', 'Neutral', 'Unsatisfied'])
plt.title('Customer Satisfaction Level Distribution', fontsize=12, fontweight='bold')
plt.xlabel('Satisfaction Level')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig('satisfaction_distribution.png', dpi=300)
plt.show()

```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap of Numerical Features')
plt.savefig('correlation_heatmap.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='Membership Type', y='Total Spend', hue='Gender', data=df, palette='Set2')
plt.title('Total Spend Distribution across Membership Types & Gender', fontsize=12, fontweight='bold')
plt.xlabel('Membership Type')
plt.ylabel('Total Spend ($)')
plt.tight_layout()
plt.savefig('spend_by_membership.png', dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.heatmap(df.isnull(), cbar=False, cmap='Blues_r', yticklabels=False)
plt.title('Missing Value Heatmap (Satisfaction Level Has 2 Missing Values)', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig('missing_values.png', dpi=300)
plt.show()
```

```python

df_clean = df.copy()

df_clean['Satisfaction Level'].fillna(df_clean[df_clean['Membership Type'] == 'Bronze']['Satisfaction Level'].mode()[0], inplace=True)

print("Remaining nulls:", df_clean.isnull().sum().sum())
```
